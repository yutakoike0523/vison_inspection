# 画像全体の放射方向FFT：image1〜image21

元のノートブックと同じFFT、ハン窓、正規化なしの放射平均、dB変換、画像全体の中央値を使用します。実際の解析セルで使っていたストライドは4です。放射平均は疎行列で集計するため、浮動小数点の加算順序が異なります。結果は画像ごとに保存し、処理済みの画像は再実行時に飛ばします。元のノートブックは `FFT_radial_full_image_original.ipynb` に保存しています。

```python
from pathlib import Path
from types import SimpleNamespace
from fft_radial_full_image_batch import run

DATA_DIR = Path("data20260903/data20260903")
OUTPUT_DIR = Path("output_radial_full_image")
WINDOW_SIZE = 128
STRIDE = 4
BATCH_SIZE = 64
FFT_WORKERS = -1
```

```python
args = SimpleNamespace(data=DATA_DIR, output=OUTPUT_DIR, window=WINDOW_SIZE, stride=STRIDE, batch_size=BATCH_SIZE, workers=FFT_WORKERS, start_number=1, end_number=21)
results = run(args)
```

```python
for item in results:
    print(Path(item["image"]).name, item["feature_shape"], f"{item['analysis_seconds']:.1f} s")
```
